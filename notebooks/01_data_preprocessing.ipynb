```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load the processed dataset
file_path = "../data/processed/processed_data.csv"

processed = pd.read_csv(file_path)

print("Processed dataset loaded!")
print("Shape:", processed.shape)

# Identify the 28 sensor features
sensor_columns = [
    col for col in processed.columns
    if col.startswith("sensor_") or col.startswith("virtual_sensor_")
]

print("Number of sensor features:", len(sensor_columns))

# Normalize sensor features
scaler = StandardScaler()

processed[sensor_columns] = scaler.fit_transform(
    processed[sensor_columns]
)

print("Sensor normalization completed!")

# Create sequences
sequence_length = 30

X_data = processed[sensor_columns].values
y_data = processed["RUL"].values

X_sequences = []
y_sequences = []

for i in range(len(processed) - sequence_length):

    X_sequences.append(
        X_data[i:i + sequence_length]
    )

    y_sequences.append(
        y_data[i + sequence_length]
    )

X_sequences = np.array(X_sequences)
y_sequences = np.array(y_sequences)

print("\nSequence creation completed!")
print("X_sequences:", X_sequences.shape)
print("y_sequences:", y_sequences.shape)
```

```python
split_index = int(len(X_sequences) * 0.80)

X_train = X_sequences[:split_index]
X_val = X_sequences[split_index:]

y_train = y_sequences[:split_index]
y_val = y_sequences[split_index:]

print("Data split completed!")

print("\nTraining:")
print("X_train:", X_train.shape)
print("y_train:", y_train.shape)

print("\nValidation:")
print("X_val:", X_val.shape)
print("y_val:", y_val.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, LSTM, Dense, Dropout

model = Sequential([
    Conv1D(
        filters=32,
        kernel_size=3,
        activation="relu",
        input_shape=(30, 28)
    ),

    MaxPooling1D(pool_size=2),

    LSTM(64),

    Dropout(0.2),

    Dense(32, activation="relu"),

    Dense(1)
])

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python
# Train the CNN-LSTM model

history = model.fit(
    X_train,
    y_train,
    validation_data=(X_val, y_val),
    epochs=10,
    batch_size=64,
    verbose=1
)

print("Training completed successfully!")
```

```python
# Generate predictions on validation data

y_pred = model.predict(X_val)

# Convert predictions from (N, 1) to (N,)
y_pred = y_pred.flatten()

print("Prediction completed!")

print("Actual RUL shape:", y_val.shape)
print("Predicted RUL shape:", y_pred.shape)

print("\nFirst 10 predictions:")
for i in range(10):
    print(
        f"Actual RUL: {y_val[i]:.2f}  |  "
        f"Predicted RUL: {y_pred[i]:.2f}"
    )
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Calculate MAE
mae = mean_absolute_error(y_val, y_pred)

# Calculate RMSE
rmse = np.sqrt(mean_squared_error(y_val, y_pred))

print("====================================")
print("MODEL EVALUATION RESULTS")
print("====================================")
print(f"MAE  : {mae:.4f}")
print(f"RMSE : {rmse:.4f}")
print("====================================")
```

```python
import numpy as np

# Sensor columns
sensor_columns = [
    col for col in processed.columns
    if col.startswith("sensor_") or col.startswith("virtual_sensor_")
]

sequence_length = 30

X_sequences_corrected = []
y_sequences_corrected = []

# Group data by A_2
for unit_id, group in processed.groupby("A_2", sort=False):

    # Get sensor values
    sensor_data = group[sensor_columns].values

    # Get RUL values
    rul_data = group["RUL"].values

    # Create sequences only within this group
    for i in range(len(group) - sequence_length):
        X_sequences_corrected.append(
            sensor_data[i:i + sequence_length]
        )

        # RUL corresponding to the final timestep
        y_sequences_corrected.append(
            rul_data[i + sequence_length - 1]
        )

# Convert to NumPy arrays
X_sequences_corrected = np.array(X_sequences_corrected, dtype=np.float32)
y_sequences_corrected = np.array(y_sequences_corrected, dtype=np.float32)

print("Corrected sequence creation completed!")
print("X shape:", X_sequences_corrected.shape)
print("y shape:", y_sequences_corrected.shape)
```

```python
from sklearn.model_selection import train_test_split

# Split corrected sequences into training and validation sets
X_train_corrected, X_val_corrected, y_train_corrected, y_val_corrected = train_test_split(
    X_sequences_corrected,
    y_sequences_corrected,
    test_size=0.2,
    random_state=42
)

print("Corrected data split completed!")

print("\nTraining data:")
print("X_train:", X_train_corrected.shape)
print("y_train:", y_train_corrected.shape)

print("\nValidation data:")
print("X_val:", X_val_corrected.shape)
print("y_val:", y_val_corrected.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv1D, MaxPooling1D, LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

# Create improved CNN-LSTM model
model_corrected = Sequential([
    Conv1D(
        filters=32,
        kernel_size=3,
        activation="relu",
        input_shape=(30, 28)
    ),

    MaxPooling1D(pool_size=2),

    LSTM(64),

    Dropout(0.3),

    Dense(32, activation="relu"),

    Dense(1)
])

# Compile model
model_corrected.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

# Display model architecture
model_corrected.summary()
```

```python
# Early stopping to prevent overfitting

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=2,
    restore_best_weights=True
)

# Train the corrected CNN-LSTM model

history_corrected = model_corrected.fit(
    X_train_corrected,
    y_train_corrected,
    validation_data=(
        X_val_corrected,
        y_val_corrected
    ),
    epochs=10,
    batch_size=64,
    callbacks=[early_stopping],
    verbose=1
)

print("Corrected CNN-LSTM training completed!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Generate predictions using the corrected model
y_pred_corrected = model_corrected.predict(X_val_corrected)

# Convert from (N, 1) to (N,)
y_pred_corrected = y_pred_corrected.flatten()

# Calculate evaluation metrics
mae_corrected = mean_absolute_error(
    y_val_corrected,
    y_pred_corrected
)

rmse_corrected = np.sqrt(
    mean_squared_error(
        y_val_corrected,
        y_pred_corrected
    )
)

print("====================================")
print("CORRECTED MODEL EVALUATION")
print("====================================")
print(f"MAE  : {mae_corrected:.4f}")
print(f"RMSE : {rmse_corrected:.4f}")
print("====================================")

print("\nCOMPARISON WITH ORIGINAL MODEL")
print("====================================")
print(f"Original MAE  : 8.9058")
print(f"Corrected MAE : {mae_corrected:.4f}")
print()
print(f"Original RMSE  : 10.3066")
print(f"Corrected RMSE : {rmse_corrected:.4f}")
print("====================================")
```

```python
# ==========================================
# STEP 22 - FUNCTION FOR RUL PREDICTION
# ==========================================

def predict_rul_for_unit(unit_id):

    # Select the requested unit
    unit_data = df[df["A_2"] == unit_id].copy()

    # Check whether unit exists
    if len(unit_data) == 0:
        print(f"Aircraft/Unit {unit_id} was not found.")
        return

    # Check whether enough observations exist
    if len(unit_data) < 30:
        print(
            f"Aircraft/Unit {unit_id} has only "
            f"{len(unit_data)} observations. "
            f"At least 30 are required."
        )
        return

    # Reset order
    unit_data = unit_data.reset_index(drop=True)

    # Take latest 30 observations
    latest_30 = unit_data[sensor_columns].tail(30)

    # Normalize
    latest_30_scaled = scaler.transform(latest_30)

    # Reshape for CNN-LSTM
    sequence = latest_30_scaled.reshape(1, 30, 28)

    # Predict
    prediction = model_corrected.predict(
        sequence,
        verbose=0
    )

    predicted_rul = float(prediction[0][0])

    # Display result
    print("\n====================================")
    print("PREDICTIVE MAINTENANCE SYSTEM")
    print("====================================")
    print(f"Aircraft/Unit ID : {unit_id}")
    print(f"Observations     : {len(unit_data)}")
    print(f"Predicted RUL    : {predicted_rul:.2f} cycles")
    print("====================================")

    return predicted_rul
```
